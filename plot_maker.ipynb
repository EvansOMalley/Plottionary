# Plot Maker

```python
import matplotlib.pyplot as plt
import numpy as np
rng = np.random.default_rng()
```

```python
def get_data():
    plots = {}
    plt_cnt = 1
    
    while True:
        try:
            with open(f"Source/data/plot{plt_cnt}.csv", 'r') as f:
                r = f.read()
                try:
                    x, y = r.split('\n')
                    x = np.array([np.float64(i) for i in x.split(',')])
                    y = np.array([np.float64(i) for i in y.split(',')])
                    plots[f"plot{plt_cnt}"] = (x, y)
                except:
                    x = np.array([np.float64(i) for i in r.split(',')])
                    plots[f"plot{plt_cnt}"] = x
            plt_cnt += 1
        except:
            break

    return plots

plots = get_data()
```

```python
def Plt_Config(title = None,
               xlab = None,
               ylab = None,
               xmin = None, xmax = None,
               ymin = None, ymax = None
              ):

    fig, ax = plt.subplots()
    fig.set_dpi(250)
    fig.set_size_inches(3, 2)
    ax.tick_params(which = "minor", labelsize = 5, size = 2)
    ax.tick_params(which = "major", labelsize = 5, size = 4)

    return (fig, ax)
```

## Plot 1

```python
fig, ax1 = Plt_Config()
p1 = plots["plot1"]
ax1.hist(p1, bins = 15)
ax1.axvline(np.mean(p1), color = "Orange")
pass
```

## Plot 2

```python
fig, ax = Plt_Config()
p2 = plots["plot2"]
ax.scatter(p2[0], p2[1], s = 0.5)
lin = np.polyfit(p2[0], p2[1], 1)
ax.axline((0, lin[1]), slope = lin[0], color = "Orange")
pass
```

## Plot 3

```python
fig, ax = Plt_Config()
p3 = plots["plot3"]
ax.hist(p3, bins = np.arange(25, 36, 1))
ax.axvline(np.percentile(p3, 85), color = "Orange")
pass
```

## Plot 4

```python
fig, ax = Plt_Config()
p4 = plots["plot4"]
ax.hist(p4, bins = np.arange(0.25, 0.75, 1/100))
ax.axvline(np.mean(p4), color = "Orange")
ax.axvline(np.mean(p4) + np.std(p4), color = "Green")
ax.axvline(np.mean(p4) - np.std(p4), color = "Green")
pass
```

## Plot 5

```python
fig, ax = Plt_Config()
p5 = plots["plot5"]
zeros = np.mean(p5 == 0)
ones = np.mean(p5 == 1)
ax.bar(x = (0,1), height = (zeros, ones), tick_label = ("Tails", "Heads"))
ax.axhline(np.mean(p4), color = "Orange")
ax.axhline(np.mean(p4) + np.std(p4), color = "Green")
ax.axhline(np.mean(p4) - np.std(p4), color = "Green")
pass
```
